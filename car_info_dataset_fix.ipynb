```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("safety_combined_dataset.csv")
```

```python
df.columns
```

```python
df.head(2)
```

```python
df.loc[df["median_price"] == 0, "median_price"] = np.nan
df.loc[df["min_price"] == 0, "min_price"] = np.nan
```

```python
df["min_price"].value_counts()
```

```python
df["median_price"].value_counts()
```

```python
# null value imputations
df['min_price'] = df['min_price'].fillna(df.groupby(['make','model','year'])['min_price'].transform('median'))
df['min_price'] = df['min_price'].fillna(df.groupby(['make','model'])['min_price'].transform('median'))
# df['min_price'] = df['min_price'].fillna(df.groupby(['make'])['min_price'].transform('median'))
```

```python
df['median_price'] = df['median_price'].fillna(df.groupby(['make','model','year'])['median_price'].transform('median'))
df['median_price'] = df['median_price'].fillna(df.groupby(['make','model'])['median_price'].transform('median'))
# df['median_price'] = df['median_price'].fillna(df.groupby(['make'])['median_price'].transform('median'))
```

```python
df['accident_safety_rating'] = df['accident_safety_rating'].fillna(df.groupby(['make','model','year'])['accident_safety_rating'].transform('median'))
df['accident_safety_rating'] = df['accident_safety_rating'].fillna(df.groupby(['make','model'])['accident_safety_rating'].transform('median'))
```

```python
print(df.min_price.isnull().sum())
print(df.median_price.isnull().sum())
print(df.accident_safety_rating.isnull().sum())
```

```python
df= df.dropna(subset=['min_price','median_price','accident_safety_rating'])
```
