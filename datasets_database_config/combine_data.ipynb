```python
import pandas as pd
import io
from kbb import Cars
import time
```

```python
carapi = pd.read_csv("carapi-opendatafeed-sample.csv")
carapi
```

```python
unique_cars = carapi[["Make Name", "Model Name", "Trim Year"]].drop_duplicates()
unique_cars
```

```python
car_prices = []
for index, row in unique_cars.iterrows():
    car = Cars(row['Make Name'].lower(), row['Model Name'].lower(), row['Trim Year'])
    car_prices.append(car.get_prices())
    #time.sleep()
```

```python
results = pd.DataFrame(car_prices)
results.to_csv("carapi-fair-prices.csv")
```

```python
results
```

```python
carapi = carapi.rename(columns={"Make Id": "make_id", "Make Name": "make", "Model Id": "model_id", "Model Name": "model", "Trim Id": "trim_id", "Trim Year": "year", "Trim Name": "trim"})
carapi
```

```python
carapi["make"] = carapi["make"].str.lower()
carapi["model"] = carapi["model"].str.lower()
```

```python
combined_dataset = pd.merge(carapi, results, on=["make", "model", "year"], how="left")
combined_dataset.to_csv("combined.csv", index=False)
```

```python
safety = pd.read_csv("safercar_data.csv")
safety = safety.rename(columns={"MAKE": "make", "MODEL": "model", "MODEL_YR": "year"})
safety["make"] = safety["make"].str.lower()
safety["model"] = safety["model"].str.lower()
safety
```

```python
safety["DRIVE_TRAIN"].value_counts()
```

```python
combined_dataset = combined_dataset.rename(columns={"Engine Drive Type": "drive_train"})
safety = safety.rename(columns={"DRIVE_TRAIN": "drive_train"})
```

```python
safety = safety.rename(columns={"OVERALL_STARS": "accident_safety_rating"})
```

```python
def drivetrain(row):
    if row["drive_train"] == "rear wheel drive":
        return "RWD"
    elif row["drive_train"] == "front wheel drive":
        return "FWD"
    elif row["drive_train"] == "all wheel drive":
        return "AWD"
    else:
        return "4WD"

combined_dataset["drive_train"] = combined_dataset.apply(drivetrain, axis=1)
```

```python
combined_dataset.drive_train.value_counts()
```

```python
combined_dataset
```

```python
safety = safety[["make", "model", "year", "drive_train", "accident_safety_rating"]]
```

```python
safety
```

```python
safety_unique = safety.drop_duplicates(subset=["make", "model", "year", "drive_train"])
```

```python
safety_combined_dataset = pd.merge(combined_dataset, safety_unique, on=["make", "model", "year", "drive_train"], how="left")
```
