```python
import pandas as pd
import io
from kbb import Cars
import time
```

```python
car_info = pd.read_csv("updated_car_info_and_safety_ratings2.csv")
car_info["image"] = None
car_info["expert_score"] = None
car_info["consumer_score"] = None

make = ""
model = ""
year = 1111
results = {}
for index, row in car_info.iterrows():
    make2 = row["make"]
    model2 = row["model"]
    year2 = row["year"]

    if make == make2 and model == model2 and year == year2:
        print("same car")
        car_info.loc[index, results.keys()] = results.values()
        continue
    else:
        print(make2, model2, year2)
        car = Cars(make2, model2, year2)
        results = car.get_car()

        car_info.loc[index, results.keys()] = results.values()
        make = make2
        model = model2
        year = year2

car_info.to_csv("combined.csv", index=False)
```

```python
images = car_info.image.unique()
images2 = {}
count = 1
for image in images:
    images2[image] = count
    count += 1

images2
```

```python
car_info["image_id"] = None

def set_image_id(row):
    row["image_id"] = images2[row["image"]]
    return row

car_info2 = car_info.apply(set_image_id, axis=1)

# car_info.to_csv("combined.csv", index=False)
car_info2
```

```python
car_info2.to_csv("combined.csv", index=False)
```

```python
all_mm = pd.read_csv("all_makes_models.csv")
all_mm
```

```python
makes = all_mm.make.unique()
models = all_mm.model.unique()
trims = all_mm.trim.unique()
fuel_types = all_mm.fuel_type.unique()
drivetrains = all_mm.drivetrain.unique()
images = all_mm.image.unique()


makes2 = {}
count = 1
for m in makes:
    makes2[m] = count
    count += 1

models2 = {}
count = 1
for m in models:
    models2[m] = count
    count += 1

trims2 = {}
count = 1
for t in trims:
    trims2[t] = count
    count += 1

fuel_types2 = {}
count = 1
for ft in fuel_types:
    fuel_types2[ft] = count
    count += 1

drivetrains2 = {}
count = 1
for dt in drivetrains:
    drivetrains2[dt] = count
    count += 1

images2 = {}
count = 1
for image in images:
    images2[image] = count
    count += 1

images2
```

```python
all_mm["make_id"] = None
all_mm["model_id"] = None
all_mm["trim_id"] = None
all_mm["fuel_type_id"] = None
all_mm["drivetrain_id"] = None
all_mm["image_id"] = None

def set_ids(row):
    row["make_id"] = makes2[row["make"]]
    row["model_id"] = models2[row["model"]]
    row["trim_id"] = trims2[row["trim"]]
    row["fuel_type_id"] = fuel_types2[row["fuel_type"]]
    row["drivetrain_id"] = drivetrains2[row["drivetrain"]]
    row["image_id"] = images2[row["image"]]
    return row

all_mm2 = all_mm.apply(set_ids, axis=1)
all_mm2
```

```python
all_mm2.to_csv("all_makes_models_complete.csv", index=False)
```
